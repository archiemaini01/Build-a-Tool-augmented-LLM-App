```python
from typing import Any, Callable, Dict, Optional
import re
import requests
import os
from getpass import getpass

# OpenRouter API setup
if not os.environ.get("OPENROUTER_API_KEY"):
    os.environ["OPENROUTER_API_KEY"] = getpass("Enter your OpenRouter API key: ")

OPENROUTER_BASE_URL = "https://openrouter.ai/api/v1"

print("OpenRouter API key configured.")
```

```python
# Module-level cache
CACHE: Any = None


def setup_cache() -> Any:
    """Initialize and assign a cache object to module-level CACHE."""
    global CACHE
    CACHE = {}
    return CACHE


def get_weather(
    city: str,
    unit: str = "C",
    fetcher= None
) -> dict:
    """
    Get current weather for a city using Open-Meteo.
    Results are cached by normalized (city, unit).
    """
    global CACHE

    if CACHE is None:
        setup_cache()

    city_key = city.strip().lower()
    unit_key = unit.strip().upper()

    if unit_key not in ("C", "F"):
        unit_key = "C"

    cache_key = (city_key, unit_key)

    # Return the exact same cached object
    if cache_key in CACHE:
        return CACHE[cache_key]

    # Allow a custom fetcher for testing
    if fetcher is not None:
        result = fetcher(city, unit_key)
        CACHE[cache_key] = result
        return result

    # Geocode city
    geo_url = "https://geocoding-api.open-meteo.com/v1/search"
    geo_params = {
        "name": city.strip(),
        "count": 1,
        "language": "en",
        "format": "json"
    }

    response = requests.get(geo_url, params=geo_params, timeout=10)
    response.raise_for_status()
    geo_data = response.json()

    results = geo_data.get("results", [])
    if not results:
        raise ValueError(f"City not found: {city}")

    latitude = results[0]["latitude"]
    longitude = results[0]["longitude"]

    # Get current weather
    weather_url = "https://api.open-meteo.com/v1/forecast"
    weather_params = {
        "latitude": latitude,
        "longitude": longitude,
        "current": "temperature_2m"
    }

    response = requests.get(
        weather_url,
        params=weather_params,
        timeout=10
    )
    response.raise_for_status()

    weather_data = response.json()
    temp_c = float(weather_data["current"]["temperature_2m"])

    # Convert Celsius to Fahrenheit if requested
    temp = temp_c if unit_key == "C" else (temp_c * 9 / 5) + 32

    result = {
        "city": city.strip(),
        "temp": float(temp),
        "unit": unit_key
    }

    CACHE[cache_key] = result
    return result


def multiply(a: float, b: float) -> float:
    """Return the product of a and b."""
    return float(a) * float(b)


def tool_registry() -> Dict[str, Callable]:
    """Return the available tools."""
    return {
        "weather": get_weather,
        "multiply": multiply
    }


def agent_route(prompt: str) -> str:
    """
    Route the prompt to weather, multiply, or none.
    """
    text = prompt.lower()

    # Weather keywords
    weather_words = [
        "weather",
        "temperature",
        "forecast",
        "degrees"
    ]

    if any(word in text for word in weather_words):
        return "weather"

    # Multiplication patterns / keywords
    multiply_patterns = [
        r"\d+(?:\.\d+)?\s*[x×*]\s*\d+(?:\.\d+)?",
        r"\bmultiply\b",
        r"\bproduct of\b",
        r"\bmultiplied by\b"
    ]

    if any(re.search(pattern, text) for pattern in multiply_patterns):
        return "multiply"

    return "none"


def agent_answer(
    prompt: str,
    tools: dict | None = None,
    cache=None
) -> str:
    """
    Use agent_route to choose the appropriate tool and return
    a short response.
    """
    global CACHE

    # Use supplied cache if provided
    if cache is not None:
        old_cache = CACHE
        CACHE = cache
    else:
        old_cache = None
        if CACHE is None:
            setup_cache()

    if tools is None:
        tools = tool_registry()

    route = agent_route(prompt)

    try:
        # WEATHER
        if route == "weather":
            match = re.search(
                r"weather\s+(?:in|at|for)\s+([A-Za-z][A-Za-z .'-]*?)(?:\s+in\s+(?:Fahrenheit|Celsius|F|C)\b|\?|$)",
                prompt,
                re.IGNORECASE
            )

            if not match:
                match = re.search(
                    r"(?:in|at|for)\s+([A-Za-z][A-Za-z .'-]*?)(?:\s+in\s+(?:Fahrenheit|Celsius|F|C)\b|\?|$)",
                    prompt,
                    re.IGNORECASE
                )

            if not match:
                return "I couldn't determine the city."

            city = match.group(1).strip()

            unit = "C"
            if re.search(r"\b(?:fahrenheit|F)\b", prompt, re.IGNORECASE):
                unit = "F"

            result = tools["weather"](
                city,
                unit=unit
            )

            return (
                f"Weather in {result['city']}: "
                f"{result['temp']:.1f}°{result['unit']}"
            )

        # MULTIPLICATION
        if route == "multiply":
            match = re.search(
                r"(-?\d+(?:\.\d+)?)\s*[x×*]\s*(-?\d+(?:\.\d+)?)",
                prompt,
                re.IGNORECASE
            )

            if not match:
                return "I couldn't find two numbers to multiply."

            a = float(match.group(1))
            b = float(match.group(2))

            result = tools["multiply"](a, b)

            return f"{a} * {b} = {result}"

        # NONE
        return "I don't know which tool to use."

    finally:
        # Restore module cache if a temporary override was used
        if cache is not None:
            CACHE = old_cache
```

```python
# Q1
c = setup_cache()
assert c is not None and isinstance(c, dict)
assert 'CACHE' in globals() and isinstance(CACHE, dict)

# Q2
setup_cache()
d1 = get_weather("Paris", "C")
d2 = get_weather("paris", "c")

assert d1 is d2
assert isinstance(d1["temp"], float)
assert d1["unit"] == "C"

d3 = get_weather("Paris", "F")
assert d3["unit"] == "F"
assert isinstance(d3["temp"], float)

# Q3
assert multiply(6, 7) == 42
assert abs(multiply(1.5, 2.0) - 3.0) < 1e-9

# Q4
reg = tool_registry()
assert isinstance(reg, dict)
assert "weather" in reg and "multiply" in reg
assert callable(reg["weather"]) and callable(reg["multiply"])

# Q5
assert agent_route("what's the weather in Jaipur?") == "weather"
assert agent_route("compute 42 x 19 please") == "multiply"
assert agent_route("say hello") == "none"

# Q6
cache = setup_cache()
tools = tool_registry()

o1 = agent_answer(
    "What's the weather in Mumbai in Fahrenheit?",
    tools=tools,
    cache=cache
)
assert "Mumbai" in o1 and "F" in o1

o2 = agent_answer(
    "Please do 12 * 8",
    tools=tools,
    cache=cache
)
assert "12.0 * 8.0 = 96.0" in o2 or "12 * 8 = 96" in o2

print("ALL PUBLIC TESTS PASSED!")
print(o1)
print(o2)
```
